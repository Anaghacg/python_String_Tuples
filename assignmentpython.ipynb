```python
# Setup string from the previous assignment
text = "Hello Zara, welcome to Python programming"

# a. Print the first character
print("First character:", text[0])

# b. Print the last character
print("Last character:", text[-1])

# c. Print the first 5 characters
print("First 5 characters:", text[:5])

# d. Print the last 11 characters
print("Last 11 characters:", text[-11:])

# e. Print the string in reverse
print("Reversed string:", text[::-1])

# f. Slice the word "Python"
# Positive indexing (for "Hello Zara, welcome to Python programming")
print("Extracted word (positive index):", text[23:29])

# Negative indexing (works even if the user's name changes length)
print("Extracted word (negative index):", text[-18:-12])
```

```python
strM = "Python beginner tutorial"

# a. Convert the sentence to uppercase
print("a. Uppercase:", strM.upper())

# b. Convert the sentence to lowercase
print("b. Lowercase:", strM.lower())

# c. Use capitalize to return the lowercased sentence to original input form
# capitalize() makes the first letter uppercase and all other letters lowercase
print("c. Capitalized:", strM.lower().capitalize())

# d. Count total occurrences of character 't'
print("d. Count of 't':", strM.count("t"))

# e. Replace "Python" with "Machine Learning"
print("e. Replaced string:", strM.replace("Python", "Machine Learning"))
```

```python
# Initial tuple creation
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples and store in "t_combine"
t_combine = tuple1 + tuple2
print("a. Concatenated tuple (t_combine):", t_combine)

# b. Repeat the elements of "t_combine" 3 times
t_repeated = t_combine * 3
print("b. Repeated 3 times:", t_repeated)

# c. Access the 3rd element from "t_combine"
print("c. 3rd element:", t_combine[2])

# d. Access the first three elements from "t_combine"
print("d. First three elements:", t_combine[:3])

# e. Access the last three elements from "t_combine"
print("e. Last three elements:", t_combine[-3:])
```
